```python
k=["apple","banana","cherry"]

k.append("orange1") # it add item to the last index in the list
print(k)
k.insert(1,"orange2") # it add item to the index we choose in  the list
print(k)
k.remove("banana") # it remove the item we choose from the list
print(k)
k.extend(["kartik", "Anjuu"]) # it add multiple items to the end of the list
print(k)
k.pop(1) # it remove the  item from the index we choose in the list
print(k)
k.pop() # it remove the last item from the list
print(k)

del k[0] # it delete the item from the index we choose in the list
print(k)
k.clear() # it clear the entire list
print(k)
```

# List Comp
List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.
Based on a list of fruits, you want a new list, containing only the fruits with the letter "a" in the name.

Without list comprehension you will have to write a for statement with a conditional test inside:

```python
# without using list comprehension
fruits = ["apple", "banana", "cherry"]
newlist = []

for x in fruits:
    if "a" in x:
        newlist.append(x)
        print(newlist)
```

```python
# with using list comprehension
# [new_item  for  item  in  iterable   if condition]

fruits = ["apple", "banana", "mango"]

newlist = [x for x in fruits if "a" in x]
print(newlist)
''' Iteration 1:

x = "apple"

Output: x डालना है → "apple"

Condition: "a" in "apple" ✔

Iteration 2:

x = "banana"

Output: "banana"

Condition: "a" in "banana" ✔

Iteration 3:

x = "mango"

Output: "mango"

Condition: "a" in "mango" ✔'''
```

```python
numbers = [1, 2, 3, 4, 5, 6]

newlist = ["even" if x % 2 == 0 else "odd" for x in numbers]


print(newlist)
```

```python
names = ["kartik", "ram", "anju", "om"]

newlist = ["long" if len(n) > 3 else "short" for n in names]

print(newlist)
```

```python
marks = [35, 80, 20, 60]

result = ["Pass" if m >= 40 else "Fail" for m in marks]

print(result)
```
