# Loop

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
copydict = thisdict.copy()
for x in copydict.copy():
 copydict.pop(x)
 print(copydict)
#“Jis dictionary par loop chal raha ho, 
# usko loop ke andar change (delete/update) nahi kar sakte.”

```

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
for x in thisdict:
  print("keys=",x) # it prints keys
  print("values=",thisdict[x]) # it prints values
  print("both=",x,thisdict[x])# prints key and value both
```

```python
# Printing all values in the dictionary
for x in thisdict.values():
  print(x)
```

```python
# Printing all keys in the dictionary
for x in thisdict.keys():
  print(x)
```

```python
# Printing both keys and values
for x, y in thisdict.items():
  print(x, y)
```

# nested 

```python
myfamily = {
  "child1" : {
    "name" : "Emil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Tobias",
    "year" : 2007
  },
  "child3" : {
    "name" : "Linus",
    "year" : 2011
  }
}

print(myfamily)
```

```python
# creating nested dictionary step by step 
child1 = {
  "name" : "Emil",
  "year" : 2004
}
child2 = {
  "name" : "Tobias",
  "year" : 2007
}
child3 = {
  "name" : "Linus",
  "year" : 2011
}

myfamily = {
  "child1" : child1,
  "child2" : child2,
  "child3" : child3
}
print(myfamily)
```

```python
print(myfamily["child2"]["name"])
```

```python
for x, obj in myfamily.items():
  print(x)

  for y in obj:
    print(y + ':', obj[y])
```
